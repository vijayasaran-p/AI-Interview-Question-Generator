```python
import os
import groq
from google.colab import userdata


# Retrieve API key from Colab Secrets
api_key = userdata.get('GROQ_API_KEY')

# Ensure the API key is available
if not api_key:
    raise ValueError("GROQ_API_KEY not found in Colab Secrets. Please add it.")

client = groq.Groq(api_key=api_key)


topic = input("Enter Interview Topic (e.g., Python, Data Science): ")

prompt = f"""
Generate 10 technical interview questions for a candidate interviewing for a {topic} role.
Provide only the questions, numbered properly from 1 to 10. Do not include introductory text.
"""


try:
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.7
    )

    # 5. Print the results
    print("\n" + "="*40)
    print(f"  SUCCESS! HERE ARE YOUR QUESTIONS FOR: {topic.upper()}")
    print("="*40 + "\n")
    print(response.choices[0].message.content)

except Exception as e:
    print("\n--- Something went wrong ---")
    print(f"Error Details: {e}")
    print("\nTip: If you see an 'Invalid API Key' error, please double-check that your key was copied correctly from Groq.")
```
